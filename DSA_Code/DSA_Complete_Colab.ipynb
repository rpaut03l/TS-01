## 1️⃣ Insertion Sort
🧠 **Mnemonic: "Pick → Scan → Shift → Drop"**

*Like sorting cards in your hand. Pick one card, scan right-to-left, shift bigger cards right, drop card in gap.*

```python
def insertion_sort(arr):
    """Pick → Scan → Shift → Drop"""
    for j in range(1, len(arr)):       # Pick each card
        key = arr[j]                    # Hold it
        i = j - 1                       # Start scanning left
        while i >= 0 and arr[i] > key:  # Bigger? Shift right!
            arr[i + 1] = arr[i]
            i -= 1
        arr[i + 1] = key                # Drop in gap
    return arr

print("Insertion Sort:", insertion_sort([5, 2, 4, 6, 1, 3]))
```

## 2️⃣ Merge Sort
🧠 **Mnemonic: "Split → Sort halves → Merge back"**

*Split pile in half until 1 each. Merge back: always pick smaller top card.*

```python
def merge_sort(arr):
    if len(arr) <= 1: return arr       # 1 item = already sorted!
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])        # Sort left half
    right = merge_sort(arr[mid:])       # Sort right half
    return merge(left, right)

def merge(left, right):
    result, i, j = [], 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    result.extend(left[i:]); result.extend(right[j:])
    return result

print("Merge Sort:", merge_sort([38, 27, 43, 3, 9, 82, 10]))
```

## 3️⃣ BST (Binary Search Tree)
🧠 **Mnemonic: "Left Less, Right moRe"**

*At each node: target smaller? Go left. Bigger? Go right.*

```python
class BSTNode:
    def __init__(self, key):
        self.key = key
        self.left = self.right = self.parent = None

class BST:
    def __init__(self): self.root = None
    
    def insert(self, key):
        z = BSTNode(key)
        y = None; x = self.root
        while x:
            y = x; x = x.left if z.key < x.key else x.right
        z.parent = y
        if not y: self.root = z
        elif z.key < y.key: y.left = z
        else: y.right = z
    
    def inorder(self, n=None, first=True):
        if first: n = self.root
        if n:
            self.inorder(n.left, False)
            print(n.key, end=' ')
            self.inorder(n.right, False)

t = BST()
for k in [15,6,18,3,7,17,20]: t.insert(k)
print("BST Inorder:", end=" "); t.inorder(); print()
```

## 4️⃣ Disjoint-Set Union (Union-Find)
🧠 **Mnemonic: "Find your leader, Unite your teams"**

*Path compression: everyone shortcuts to the boss. Union by rank: smaller tree under bigger.*

```python
class DSU:
    def __init__(self, n):
        self.parent = list(range(n)); self.rank = [0]*n
    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression!
        return self.parent[x]
    def union(self, x, y):
        rx, ry = self.find(x), self.find(y)
        if rx == ry: return False
        if self.rank[rx] < self.rank[ry]: rx, ry = ry, rx
        self.parent[ry] = rx
        if self.rank[rx] == self.rank[ry]: self.rank[rx] += 1
        return True

dsu = DSU(6)
dsu.union(0,1); dsu.union(2,3); dsu.union(0,2)
print(f"0 and 3 same group? {dsu.find(0)==dsu.find(3)}")  # True
print(f"0 and 4 same group? {dsu.find(0)==dsu.find(4)}")  # False
```

## 5️⃣ BFS & DFS
🧠 **BFS: "Brothers first, Sons later" (Queue)**
🧠 **DFS: "Dive First, Surface later" (Recursion/Stack)**

```python
from collections import deque

def bfs(graph, source):
    dist = {source: 0}; parent = {source: None}
    queue = deque([source])
    while queue:
        u = queue.popleft()
        for v in graph.get(u, []):
            if v not in dist:
                dist[v] = dist[u] + 1; parent[v] = u; queue.append(v)
    return dist, parent

g = {'A':['B','C'], 'B':['A','D'], 'C':['A','D'], 'D':['B','C']}
dist, par = bfs(g, 'A')
print(f"BFS distances from A: {dist}")

# Shortest path
path = []; c = 'D'
while c: path.append(c); c = par.get(c)
print(f"Shortest A→D: {path[::-1]}")
```

## 6️⃣ Dijkstra's Algorithm
🧠 **Mnemonic: "Always pick the Daintiest (smallest) Distance"**

*Greedy: visit cheapest unvisited. Relax: "Can I get there cheaper through here?"*

```python
import heapq

def dijkstra(graph, source):
    dist = {v: float('inf') for v in graph}; parent = {v: None for v in graph}
    dist[source] = 0; heap = [(0, source)]; visited = set()
    while heap:
        d_u, u = heapq.heappop(heap)
        if u in visited: continue
        visited.add(u)
        for v, w in graph[u]:
            if v not in visited and dist[u]+w < dist[v]:
                dist[v] = dist[u]+w; parent[v] = u
                heapq.heappush(heap, (dist[v], v))
    return dist, parent

g = {'A':[('B',4),('C',2)], 'B':[('A',4),('C',1),('D',5)],
     'C':[('A',2),('B',1),('E',3)], 'D':[('B',5),('E',2)], 'E':[('C',3),('D',2)]}
dist, _ = dijkstra(g, 'A')
print(f"Dijkstra from A: {dist}")
```

## 7️⃣ Max Flow (Edmonds-Karp)
🧠 **Mnemonic: "Find path → Push flow → Update pipes → Repeat"**

```python
from collections import deque

def max_flow(cap, s, t, n):
    res = [row[:] for row in cap]; total = 0
    while True:
        par = [-1]*n; par[s]=s; q = deque([s])
        while q and par[t]==-1:
            u = q.popleft()
            for v in range(n):
                if par[v]==-1 and res[u][v]>0: par[v]=u; q.append(v)
        if par[t]==-1: break
        d=float('inf'); v=t
        while v!=s: d=min(d,res[par[v]][v]); v=par[v]
        v=t
        while v!=s: u=par[v]; res[u][v]-=d; res[v][u]+=d; v=u
        total+=d
    return total

cap = [[0,10,5,0],[0,0,6,8],[0,0,0,7],[0,0,0,0]]
print(f"Max flow = {max_flow(cap,0,3,4)}")
```

## 8️⃣ Rod Cutting (DP)
🧠 **Mnemonic: "DP = Don't recompute, just look it uP!"**

*Try all first cuts. Remember best for smaller rods. Build small→big.*

```python
def rod_cutting(prices, n):
    r = [0]*(n+1); s = [0]*(n+1)
    for j in range(1, n+1):
        best = float('-inf')
        for i in range(1, j+1):
            if prices[i]+r[j-i] > best:
                best = prices[i]+r[j-i]; s[j] = i
        r[j] = best
    cuts = []; rem = n
    while rem > 0: cuts.append(s[rem]); rem -= s[rem]
    return r[n], cuts

prices = [0, 1, 5, 8, 9, 10, 17, 17, 20, 24, 30]
for n in [4, 7, 10]:
    rev, cuts = rod_cutting(prices, n)
    print(f"Rod n={n}: revenue=${rev}, cuts={cuts}")
```

## 9️⃣ LCS — Longest Common Subsequence (DP)
🧠 **Mnemonic: "Match → Diagonal+1. Miss → Max(Up, Left)"**

```python
def lcs(X, Y):
    m, n = len(X), len(Y)
    c = [[0]*(n+1) for _ in range(m+1)]
    for i in range(1, m+1):
        for j in range(1, n+1):
            if X[i-1]==Y[j-1]: c[i][j] = c[i-1][j-1]+1
            else: c[i][j] = max(c[i-1][j], c[i][j-1])
    # Reconstruct
    res = []; i, j = m, n
    while i>0 and j>0:
        if X[i-1]==Y[j-1]: res.append(X[i-1]); i-=1; j-=1
        elif c[i-1][j]>=c[i][j-1]: i-=1
        else: j-=1
    return c[m][n], ''.join(reversed(res))

length, seq = lcs("ABCBDAB", "BDCABA")
print(f"LCS of 'ABCBDAB' & 'BDCABA': '{seq}' (length {length})")
```

## 🔟 Greedy Activity Selection
🧠 **Mnemonic: "Greedy = Grab what Goes away first"**

*Sort by FINISH time. Always pick earliest-finishing compatible activity.*

```python
def activity_selection(activities):
    s = sorted(activities, key=lambda x: x[1])  # Sort by finish!
    selected = [s[0]]; last = s[0][1]
    for start, finish in s[1:]:
        if start >= last:
            selected.append((start, finish)); last = finish
    return selected

acts = [(1,4),(3,5),(0,6),(5,7),(3,9),(5,9),(6,10),(8,11),(8,12),(2,14),(12,16)]
result = activity_selection(acts)
print(f"Selected {len(result)} activities: {result}")
```
